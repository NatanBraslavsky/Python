```python
import numpy as np

# ---------- Atividade 1 ----------

# nivel 1
pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])
faturamento = pecas_vendidas * 50
dias_pico = pecas_vendidas[pecas_vendidas > 200]

print("Pecas vendidas:", pecas_vendidas)
print("Faturamento estimado:", faturamento)
print("Dias de pico (> 200):", dias_pico)

# nivel 2
vendas = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330],
])

total_por_loja = np.sum(vendas, axis=1)
media_geral = np.nanmean(vendas)
meta = np.where(vendas >= 200, 'Meta Atingida', 'Abaixo')

print("\nTotal por loja:", total_por_loja)
print("Media diaria da rede:", media_geral)
print("Meta:\n", meta)

# nivel 3
rng = np.random.default_rng(42)
departamentos = rng.choice(['Eletrônicos', 'Roupas', 'Casa'], size=50)
nomes, contagens = np.unique(departamentos, return_counts=True)

campanhas = np.array([12000, 45000, 23000, 89000, 31000])
campeã = np.argmax(campanhas)

print("\nVendas por departamento:")
for nome, qtd in zip(nomes, contagens):
    print(f"  {nome}: {qtd}")
print("Campanha campeã: indice", campeã, "com R$", campanhas[campeã])

# ---------- Atividade 2 ----------

# nivel 1
dias = np.arange(1, 31)
metas = np.linspace(20000, 30000, 5)
estoque = np.array([10.5, 20.1, 30.9]).astype(int)
ultimos_dias = dias[-5:][::-1]

print("\nDias do mes:", dias)
print("Metas:", metas)
print("Estoque (int):", estoque)
print("Ultimos 5 dias invertidos:", ultimos_dias)
```
