```python
import pandas as pd
from sqlalchemy import create_engine, text

engine = create_engine('sqlite:///sistema_rh.db')

with engine.begin() as conn:
    conn.execute(text("DROP TABLE IF EXISTS funcionarios"))
    conn.execute(text("""
        CREATE TABLE funcionarios (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            nome TEXT NOT NULL,
            cargo TEXT NOT NULL,
            salario REAL NOT NULL
        )
    """))

novo_funcionario = {"nome": "Maria Souza", "cargo": "Analista de RH", "salario": 4500.00}

with engine.begin() as conn:
    conn.execute(
        text("INSERT INTO funcionarios (nome, cargo, salario) VALUES (:nome, :cargo, :salario)"),
        novo_funcionario
    )

with engine.connect() as conn:
    df = pd.read_sql_query("SELECT * FROM funcionarios", conn)

print(df)
```
