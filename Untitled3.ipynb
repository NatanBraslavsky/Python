```python
import streamlit as st
import pandas as pd

COL_DATA = 'data'
COL_CATEGORIA = 'categoria'
COL_RECEITA = 'receita'
COL_PEDIDO = 'id_pedido'

st.title('Dashboard de Vendas')


@st.cache_data
def carregar_dados():
    return pd.read_csv('vendas.csv', parse_dates=[COL_DATA])


df = carregar_dados()

st.sidebar.title('Filtros')

lista_de_categorias = sorted(df[COL_CATEGORIA].unique())
categorias_selecionadas = st.sidebar.multiselect(
    'Selecione as Categorias',
    options=lista_de_categorias,
    default=lista_de_categorias,
)

df_filtrado = df[df[COL_CATEGORIA].isin(categorias_selecionadas)]

if df_filtrado.empty:
    st.warning('Nenhuma categoria selecionada. Escolha ao menos uma no painel lateral.')
    st.stop()

receita_calculada = df_filtrado[COL_RECEITA].sum()
total_pedidos = df_filtrado[COL_PEDIDO].nunique()

col1, col2 = st.columns([1, 1])

with col1:
    st.metric(label='Receita Total', value=f'R$ {receita_calculada:,.2f}')

with col2:
    st.metric(label='Total de Pedidos', value=f'{total_pedidos:,}')

aba1, aba2 = st.tabs(['Evolução Mensal', 'Tabela de Dados'])

with aba1:
    dados_agrupados = (
        df_filtrado
        .groupby(df_filtrado[COL_DATA].dt.to_period('M').dt.to_timestamp())[COL_RECEITA]
        .sum()
    )
    st.area_chart(dados_agrupados)

with aba2:
    st.dataframe(df_filtrado)

    st.download_button(
        label='Baixar recorte em CSV',
        data=df_filtrado.to_csv(index=False).encode('utf-8'),
        file_name='vendas_filtradas.csv',
        mime='text/csv',
    )
```
